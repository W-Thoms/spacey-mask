This function is to plot the confusion matrix.


```python
def plot_confusion_matrix(y,y_predict):
    "this function plots the confusion matrix"
    from sklearn.metrics import confusion_matrix

    cm = confusion_matrix(y, y_predict)
    ax= plt.subplot()
    sns.heatmap(cm, annot=True, ax = ax); #annot=True to annotate cells
    ax.set_xlabel('Predicted labels')
    ax.set_ylabel('True labels')
    ax.set_title('Confusion Matrix'); 
    ax.xaxis.set_ticklabels(['did not land', 'land']); ax.yaxis.set_ticklabels(['did not land', 'landed']) 
    plt.show() 
```

## Below is my own analysis of the tree model since it is highly variable.

The notebook tasks above create a tree model that varies highly between runs because the test sample is so small and the decision tree model creates a different tree with different outcomes each time. I observed this variance between .66 and .85 accuracy with cofusion matrices that had widely varied outcomes. The cells below attempt to repeatedly run the decision tree and either take its average score and average confusion matrix or summed confusion matrix to get a better idea of how accurate it is on repeated runs. 

```python
scores = []
for seed in range(20):
    tree = DecisionTreeClassifier(random_state=seed)
    tree_cv = GridSearchCV(tree, parameters, cv=10)
    tree_cv.fit(X_train, Y_train)
    score = tree_cv.score(X_test, Y_test)
    scores.append(score)

avg_tree_score = np.mean(scores)
std_tree_score = np.std(scores)
print(f"Average test accuracy over 20 runs: {avg_tree_score:.4f} (+/- {std_tree_score:.4f})")
```

```python
from sklearn.model_selection import RepeatedStratifiedKFold

cv_strategy = RepeatedStratifiedKFold(n_splits=10, n_repeats=5, random_state=2)

tree = DecisionTreeClassifier(random_state=2)
tree_cv = GridSearchCV(tree, parameters, cv=cv_strategy)
tree_cv.fit(X_train, Y_train)

print(tree_cv.best_score_)
```

```python
from sklearn.metrics import confusion_matrix

cm_list = []
scores = []

for seed in range(20):
    tree = DecisionTreeClassifier(random_state=seed)
    tree_cv = GridSearchCV(tree, parameters, cv=10)
    tree_cv.fit(X_train, Y_train)
    
    yhat = tree_cv.predict(X_test)
    score = tree_cv.score(X_test, Y_test)
    cm = confusion_matrix(Y_test, yhat)
    
    scores.append(score)
    cm_list.append(cm)

avg_tree_score = np.mean(scores)
avg_cm = np.mean(cm_list, axis=0)  # average confusion matrix across all runs

print(f"Average test accuracy: {avg_tree_score:.4f}")
print("Average confusion matrix:\n", avg_cm)

ax = plt.subplot()
sns.heatmap(avg_cm, annot=True, fmt='.2f', ax=ax)  # fmt='.2f' since values are now averaged floats
ax.set_xlabel('Predicted labels')
ax.set_ylabel('True labels')
ax.set_title('Average Confusion Matrix (20 runs)')
ax.xaxis.set_ticklabels(['did not land', 'land'])
ax.yaxis.set_ticklabels(['did not land', 'landed'])
plt.show()
```

```python
Sum_tree_score = np.sum(scores)
Sum_cm = np.sum(cm_list, axis=0)  # average confusion matrix across all runs

print(f"Sum test accuracy: {Sum_tree_score:.4f}")
print("Sum confusion matrix:\n", Sum_cm)

ax = plt.subplot()
sns.heatmap(Sum_cm, annot=True, fmt='.2f', ax=ax)  # fmt='.2f' since values are now averaged floats
ax.set_xlabel('Predicted labels')
ax.set_ylabel('True labels')
ax.set_title('Sum Confusion Matrix (20 runs)')
ax.xaxis.set_ticklabels(['did not land', 'land'])
ax.yaxis.set_ticklabels(['did not land', 'landed'])
plt.show()
```
